# Nested loop

# Increasing Triangle

```python
for i in range(1,6):
  for j in range(1,i+1):
    print("*" , end=" ")
  print()
```

# Decreasing Triangle

```python
for i in range(5, 0, -1):

    for j in range(i):

        print("*", end=" ")

    print()
```

# Increasing with numbers

```python
for i in range(1,6):
  for i in range(1,i+1):
    print(i, end=" ")
  print()
```

# Decreasing with numbers

```python
for i in range(5, 0, -1):

    for j in range(1, i + 1):

        print(j, end=" ")

    print()
```

# Square pattern

```python
for i in range(1,6):
  for j in range(1,6):
    print("*",end=" ")
  print()
```

# Right-Aligned Increasing Triangle

```python
for i in range(1,6):
  for j in range(5-i):
    print(" ",end=" ")
  for j in range(i):
    print("*",end=" ")
  print()
```

# Right-Aligned Decreasing Triangle

```python
for i in range(5,0,-1):
  for j in range(5-i):
    print(" ",end=" ")
  for j in range(i):
    print("*",end=" ")
  print()
```

#  Functions: Local vs. Global Variables

```python
count=0
def increment():
  count=0
  count+=1
  print("In function(local):",count)
increment()
print("out fimction(global):",count)
```

```python
count=0
def increment():
  global count
  count+=1
  print("In function:",count)
increment()
print("out fimction:",count)
```

```python
increment()
print("out fimction:",count)
```
