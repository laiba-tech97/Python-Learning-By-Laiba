# Shopping Cart

```python
cart_items = ["shampoo", "soap", "toothpaste"]
prices = [250, 60, 120]
for i,item in enumerate(cart_items):
  print(f"{i+1}.{item} Rs {prices[i]}")
```

```python
print("Total=",sum(prices))
```

```python
cart_items = ["shampoo", "soap", "toothpaste"]
prices = [250, 60, 120]
item = input("Which item do you want to remove? ")
if item in cart_items:
    index = cart_items.index(item)
    cart_items.pop(index)
    prices.pop(index)
print("Updated cart:", cart_items)
print("Updated prices:", prices)
```

# Working with Two Lists Together

```python
names = ["Ali", "Sara", "Bilal"]
scores = [85, 92, 76]
for names,scores in zip(names,scores):
  print(f"{names} scored {scores}")
```

```python
names = ["Ali", "Sara", "Bilal"]
scores = [85, 92, 76]
high=max(scores)
index=scores.index(high)
print("High score=",names[index])
```

```python
pass_fail=[]
for score in scores:
  if score >= 50:
    pass_fail.append("Pass")
  else:
    pass_fail.append("Fail")
print(pass_fail)
```

# Grade Analyzer

```python
grades = [45, 67, 89, 32, 76, 90, 55, 40]
print("Fail grades:")
for grade in grades:
    if grade < 50:
        print(grade)
print("Distinction grades:")
for grade in grades:
    if grade >= 85:
        print(grade)
grades.sort(reverse=True)
print(grades)
print("Top 3 grades:", grades[:3])
average = sum(grades) / len(grades)
print("average:", average)
passed = 0
for grade in grades:
    if grade >= 50:
        passed += 1
print("Students passed:", passed, "out of", len(grades))
```
