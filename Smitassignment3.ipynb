# Pattern Printing

```python
# Right-angle triangle
for i in range(1,6):
  for j in range(1,i+1):
    print("*" , end=" ")
  print()
```

```python
#Right-angle triangle (reverse)
for i in range(5,0,-1):
  for j in range(5-i):
    print(" ",end=" ")
  for j in range(i):
    print("*",end=" ")
  print()
```

```python
n = 5
# Upper part
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()
# Lower part
for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()
```

```python
#Square pattern
for i in range(1,6):
  for j in range(1,6):
    print("*",end=" ")
  print()
```

# Simple Calculator

```python
def add(a,b):
  return a+b
def sub(a,b):
  return a-b
def multiply(a,b):
  return a*b
def division(a,b):
  return a/b
try:
  num1=int(input("Enter 1st no:"))
  num2=int(input("Enter 2nd no:"))
  operator=input("Enter which operation to perform (*,/,+,-)")
  if operator=="+":
    result=add(num1,num2)
    print("Addition of numbers = ",result)
  elif operator=="-":
    result=sub(num1,num2)
    print("Subtraction of numbers = ",result)
  elif operator=="*":
    result=multiply(num1,num2)
    print("Multiplication of numbers = ",result)
  elif operator=="/":
    result=division(num1,num2)
    print("Division of numbers = ",result)
  else:
      print("Invalid choice")
except ValueError:
  print("Enter numbers only greater than 0")
```

# Student Info System

```python
students = {
    "Ali": {"id": "101","marks": 85},
    "Sara": {"id": "102","marks": 92},
    "Laiba": {"id": "103","marks": 88 }
}
search = input("Enter Name or ID: ")
if search in students:
    print("Name:", search)
    print("ID:", students[search]["id"])
    print("Marks:", students[search]["marks"])
else:
    found = False
    for name in students:
        if students[name]["id"] == search:
            print("Name:", name)
            print("ID:", students[name]["id"])
            print("Marks:", students[name]["marks"])
            found = True
            break
    if found == False:
        print("Student not found")
```

# Duplicate Word Remover

```python
sentence = input("Enter a sentence: ")
def remove_duplicates(sentence):
    words = sentence.split()
    unique_words = set()
    for word in words:
        unique_words.add(word)
    return words, unique_words
words, unique_words = remove_duplicates(sentence)
print("Unique words:", list(unique_words))
print("Duplicates removed:", len(words) - len(unique_words))
```

# Word Frequency Counter

```python
text = input("Enter a paragraph: ")
def count_words(text):
    words = text.lower().split()
    count = {}
    for word in words:
        if word in count:
            count[word] += 1
        else:
            count[word] = 1
    return count
result = count_words(text)
sorted_words = sorted(result.items(), key=lambda x: x[1], reverse=True)
for word, frequency in sorted_words:
    print(word, ":", frequency)
```
